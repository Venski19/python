```python
def tabuada(numero):
    for contador in range(1, 11):  # Gera números de 1 a 10
        print(f"{numero} // {contador} = {numero // contador}")
numero = int(input("Digite um número para ver sua tabuada: "))
tabuada(numero)
```

#**AULA 7** ⛽ ➖ ⛰ listas

```python
import random
dica = [
    "🚀O começo é agora: Transforme a inércia em impulso e dê o seu melhor primeiro passo",
    "💡Seja a energia que você quer atrair. Mantenha o ritmo e a consistência",
    "✨Chegamos à metade! Respire fundo, ajuste a rota se preciso e mantenha o foco na meta",
    "💪A persistência é a chave. Lembre-se do porquê você começou e avance com determinação",
    "🏆Seu esforço da semana merece reconhecimento. Finalize com excelência e prepare-se para recarregar",
    "🧭Não é sobre a velocidade, mas sobre a direção. Aproveite para refletir e planejar a próxima jornada",
    "💖Acalme a mente e recarregue o espírito. Um novo ciclo de oportunidades espera por você amanhã",

]
while True:
  pedido = input("Digite 'dica' para receber uma dica ou 'sair' para encerrar \n").lower()

  if pedido == "dica":
    dica = random.choice(dica) # escolhe uma dica aleatória sair
    print(f"dica: {dica} \n")
  elif pedido == "sair":
    print("até a próxima! \n")
    break
  else:
    print("omando inválido. tente novmente. \n")
    break
```
